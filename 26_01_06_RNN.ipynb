**RNN**

**CBOW의 한계**

**1. 단어의 순서 무시 (Bag-of-Words의 한계)**
CBOW는 주변 단어들의 벡터를 추출한 뒤 이를 '더하거나(sum)' 혹은 **'평균(average)'**을 낸다.

산술적인 합산 과정에서 단어의 위치 정보가 완전히 사라진다. 예를 들어, "I eat an apple"과 "An apple eats me"는 CBOW 입장에서 맥락 벡터값이 동일하게 계산될 수 있다.

**2.고정된 윈도우 크기 (Fixed Window Size)**
CBOW는 모델을 설계할 때 윈도우 크기(예: 앞뒤 2단어)를 고정한다.만약 특정 단어를 이해하기 위해 10단어 앞의 정보가 결정적이라면, 윈도우 크기 밖의 정보는 모델이 전혀 참고할 수 없다. 즉, **장기 의존성(Long-term dependency)**을 다루는 데 구조적인 한계가 있다.

**3.연결(Concatenate) 방식의 비용 문제**
순서를 고려하기 위해 벡터를 더하지 않고 옆으로 이어 붙이는 방식($W_{in}$ 연결)을 생각할 수 있다.

문제점: 윈도우 크기가 $N$일 때, 입력층에서 은닉층으로 가는 가중치 매개변수의 수가 $N$배로 늘어난다. 맥락이 길어질수록 가중치 행렬의 크기가 비대해져 메모리 점유율이 높아지고 계산 속도가 현저히 느려진다.

**"순서도 기억하면서(1번 해결), 가변적인 길이의 문장도 처리하고(2번 해결), 가중치를 공유해서 효율도 높이는(3번 해결)" 모델인 RNN이 등장하게 된 것**
