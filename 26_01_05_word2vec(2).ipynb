인스턴스 변수에 관해 알게된 점

python에서 인스턴스 변수는 init 메서드에서 만들지 않고 다른 함수에서 만드는 것도 가능하다.

다른 함수에서 self를 인수로 받지 않아도 무방하나, 그 class 내부에 있을 이유가 없는 무의미한 역할일 확률이 높다.

그럼, 다른 함수에서 새로운 인스턴스 변수를 생성하는것이 가능하냐 > ok
하지만, 아래와 같은 문제가 있다.

1.  AttributeError 방지 (중요)

만약 forward 메서드에서 self.x를 만들고, backward에서 self.x를 사용한다고 가정해 봅시다. 그런데 누군가 실수로 forward를 호출하지 않고 backward를 먼저 호출하면 어떻게 될까요? 파이썬은 **"그런 변수(x)는 없는데?"**라며 프로그램이 멈춰버립니다(AttributeError).

2. 가독성과 유지 보수

다른 개발자(혹은 미래의 나)가 코드를 볼 때, __init__만 확인하면 **"아, 이 클래스는 이런 데이터를 가지고 있구나"**라고 한눈에 파악할 수 있습니다. 변수가 여기저기서 갑자기 튀어나오면 코드를 추적하기 매우 힘들어집니다.

3. 딥러닝 프레임워크와의 호환성
나중에 PyTorch나 TensorFlow 같은 라이브러리를 쓰게 되면, 모델의 가중치나 레이어는 반드시 초기화 단계(__init__)에서 등록되어야 시스템이 "이것들이 학습해야 할 대상이구나"라고 인식하고 추적할 수 있습니다.
